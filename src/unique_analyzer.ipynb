```python
import csv
from io import StringIO
from collections import Counter

def analyze_calendar_data():
    """
    Parses the provided CSV data, processes the 'Items' field into sorted lists,
    and calculates the total number of doors vs. the number of distinct item sets.
    """
    # The provided CSV data as a multi-line string
    csv_data = """Door Number,Items,Refrigerated,Special Day
1,"Pingui,Choco Fresh,Maxi King",True,first advent 
2,"Duplo,None,Hazelnut Kugel",False,
3,"Bueno,None,Crispy Bits Kugel",False,
4,"Duplo,Hazelnut Kugel,Crispy Bits Kugel",False,
5,"Cards,None,Crispy Bits Kugel",False,
6,"Nikolaus,Überraschung,None",False,Nikolaus
7,"Choco Fresh,Pingui,Maxi King",True,second advent
8,"Duplo,Hazelnut Kugel,Crispy Bits Kugel",False,
9,"None,Bueno,Hazelnut Kugel",False,
10,"None,Duplo,Crispy Bits Kugel",False,
11,"Bueno,Hazelnut Kugel,Crispy Bits Kugel",False,
12,"Choco Fresh,Pingui,None",True,halftime
13,"Hazelnut Kugel,Duplo,Happy Hippo",False,
14,"Choco Fresh,Maxi King,None",True,third advent
15,"Cards,Bueno,Hazelnut Kugel",False,
16,"Duplo,Crispy Bits Kugel,Cards",False,
17,"Bueno,Hazelnut Kugel,Happy Hippo",False,
18,"Duplo,Happy Hippo,Crispy Bits Kugel",False,
19,"Duplo,Crispy Bits Kugel,Hazelnut Kugel",False,
20,"Crispy Bits Kugel,Happy Hippo,Bueno",False,
21,"Choco Fresh,Pingui,None",True,fourth advent
22,"Duplo,Cards,Hazelnut Kugel",False,
23,"Duplo,Crispy Bits Kugel,Happy Hippo",False,
24,"Hazelnut Kugel,Überraschung,Cards",False,christmas
"""

    # Use StringIO to treat the string data like a file
    csvfile = StringIO(csv_data)
    
    # Use the DictReader for easy access by column header
    reader = csv.DictReader(csvfile)
    
    # List to store processed door data
    processed_data = []
    # List to collect all valid items for individual item analysis
    all_items = []
    # Set to store unique item combinations (as tuples)
    distinct_item_sets = set()
    # Counter to track the frequency of each distinct set
    item_set_counts = Counter()

    print("--- Processed Calendar Data (Door Number & Sorted Items) ---\n")

    for row in reader:
        # 1. Parse the 'Items' field and clean up
        raw_items = [item.strip() for item in row['Items'].split(',')]
        
        # 2. Filter out 'None' placeholder items
        valid_items = [item for item in raw_items if item != 'None' and item]
        
        # 3. Sort the list alphabetically to ensure order doesn't affect uniqueness
        valid_items.sort()
        
        # 4. Create a tuple representation for the item set for use in a set/Counter
        item_set_tuple = tuple(valid_items)
        
        # Track the unique set and its count
        distinct_item_sets.add(item_set_tuple)
        item_set_counts[item_set_tuple] += 1
        
        # Prepare the structured entry
        door_data = {
            'Door Number': int(row['Door Number']),
            'Items': valid_items,
            'Refrigerated': row['Refrigerated'].lower() == 'true',
            'Special Day': row['Special Day']
        }
        
        processed_data.append(door_data)
        all_items.extend(valid_items)
        
        # Print a summary of the processed door
    
    # --- Final Item Set Analysis ---
    
    total_doors = len(processed_data)
    count_distinct_sets = len(distinct_item_sets)
    
    print("\n" + "="*40)
    print("--- Item Set Combination Analysis ---")
    print("="*40)
    
    print(f"Total number of doors in the calendar: {total_doors}")
    print(f"Number of DISTINCT item combinations (sets) found: {count_distinct_sets}")
    print(f"Number of item combinations that are REPEATED: {total_doors - count_distinct_sets}")

    print("\nBreakdown of Each Distinct Item Set and its Frequency:")
    # Sort the output by frequency (most common first)
    for item_set, count in item_set_counts.most_common():
        # Displaying the list representation of the set
        set_display = list(item_set) if item_set else ['(Empty)']
        print(f" - Combination: {set_display} | Count: {count}")
        
    # --- Original Individual Item Analysis (kept for context) ---
    
    total_items = len(all_items)
    distinct_items = set(all_items)
    count_distinct_items = len(distinct_items)
    item_counts = Counter(all_items)

    print("\n" + "="*40)
    print("--- Individual Item Analysis ---")
    print("="*40)
    print(f"Total number of individual item slots opened: {total_items}")
    print(f"Number of distinct item types: {count_distinct_items}")
    
    print("\nIndividual Item Breakdown:")
    for item, count in sorted(item_counts.items()):
        print(f"- {item}: {count} appearances")

```

```python
analyze_calendar_data()
```
