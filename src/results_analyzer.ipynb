```python
import pandas as pd
```

```python
available_products_file_name = '../data/availableProducts.csv'
available_products_df = pd.read_csv(available_products_file_name, index_col='Product name')
available_products_df.head()
```

```python
assignment_file_name = '../results/door_assignment_list.csv'
door_assignment_df = pd.read_csv(assignment_file_name)
door_assignment_df.head()
```

```python
product_quantity_df = door_assignment_df[['Item1', 'Item2', 'Item3']].apply(pd.Series.value_counts)
product_quantity_df['Quantity'] = product_quantity_df.sum(axis=1)
product_quantity_df['Packaging quantity'] = available_products_df.loc[product_quantity_df.index, 'minimum package quantity']
product_quantity_df.drop([f'Item{index}' for index in range(1,3+1)], axis=1, inplace=True)
product_quantity_df["Bought"] = [True, True, False, True, False, False, False, False, False, False]
product_quantity_df
```
